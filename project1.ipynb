# Part C: mapping

```python
# Part C - Index Reference Genome and Map Reads
"""
Step C1-C2: Index reference genome and map Illumina short-reads with BWA
"""

print("="*60)
print("PART C: READ MAPPING WITH BWA")
print("="*60)

import os

ref_genome = "Project1/input_data/reference_genome.fasta"
fastq_file = "Project1/input_data/SRR36584168.fastq"

# Verify files exist
if not os.path.exists(ref_genome):
    print(f"❌ Reference genome not found: {ref_genome}")
elif not os.path.exists(fastq_file):
    print(f"❌ FASTQ file not found: {fastq_file}")
else:
    # Step C1: Index the reference genome
    print("\n" + "="*40)
    print("STEP C1: Indexing reference genome with BWA")
    print("="*40)
    !bwa index {ref_genome}

    # Check index files
    print("\n✓ Reference genome indexed:")
    !ls -lh {ref_genome}.*

    # Step C2: Map reads to reference genome
    print("\n" + "="*40)
    print("STEP C2: Mapping reads with BWA-MEM")
    print("="*40)
    print(f"Mapping single-end reads to reference...")

    !bwa mem \
      -t 4 \
      -R '@RG\tID:phage\tSM:Klebsiella_phage\tLB:lib1\tPL:ILLUMINA' \
      {ref_genome} \
      {fastq_file} \
      > Project1/mapping_output/aligned_reads.sam

    # Check SAM file
    sam_file = "Project1/mapping_output/aligned_reads.sam"
    if os.path.exists(sam_file):
        file_size = os.path.getsize(sam_file) / (1024**2)
        print(f"\n✓ Mapping completed!")
        print(f"  SAM file: {sam_file}")
        print(f"  Size: {file_size:.2f} MB")
    else:
        print("❌ Mapping failed!")
```

```python
# Part C - Print SAM File Head

print("="*60)
print("PART C: SAM FILE ANALYSIS")
print("="*60)

sam_file = "Project1/mapping_output/aligned_reads.sam"

    print("\n" + "="*40)
    print("HEAD OF SAM FILE (First 15 lines)")
    print("="*40)
    !head -n 15 {sam_file}

    import subprocess
    import re

    # Extract header lines
    print("\nSAM HEADER LINES:")
    print("-" * 50)
    !grep '^@' {sam_file} | head -3

    print("\n  @HD - Header line: VN=version, SO=sorting order")
    print("  @SQ - Reference sequence: SN=reference name, LN=length")
    print("  @PG - Program: ID=program ID, PN=program name, VN=version")

    # Extract first alignment line
    cmd = f"grep -v '^@' {sam_file} | head -n 1"
    result = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    first_read = result.stdout.strip().split('\t')

    if first_read and len(first_read) >= 11:
        print("\nFIRST ALIGNMENT LINE (SAM fields):")
        print("-" * 80)

        fields = [
            ("QNAME", "Read ID", first_read[0]),
            ("FLAG", "Bitwise flag", first_read[1]),
            ("RNAME", "Reference name", first_read[2]),
            ("POS", "1-based position", first_read[3]),
            ("MAPQ", "Mapping quality", first_read[4]),
            ("CIGAR", "CIGAR string", first_read[5]),
            ("RNEXT", "Mate reference", first_read[6]),
            ("PNEXT", "Mate position", first_read[7]),
            ("TLEN", "Template length", first_read[8]),
            ("SEQ", "Sequence", first_read[9][:50] + "..." if len(first_read[9]) > 50 else first_read[9]),
            ("QUAL", "Quality", first_read[10][:50] + "..." if len(first_read[10]) > 50 else first_read[10])
        ]

        for field, desc, value in fields:
            print(f"{field:8} ({desc:20}): {value}")

        # Interpret FLAG
        flag = int(first_read[1])
        print(f"\nFLAG {flag} interpretation:")
        if flag & 1: print("  • Paired read (1) - NO (single-end data)")
        if flag & 2: print("  • Properly paired (2) - NO (single-end)")
        if flag & 4: print("  • Read unmapped (4)")
        if flag & 8: print("  • Mate unmapped (8)")
        if flag & 16: print("  • Read reverse strand (16)")
        if flag & 32: print("  • Mate reverse strand (32)")
        if flag & 64: print("  • First in pair (64)")
        if flag & 128: print("  • Second in pair (128)")
        if flag & 256: print("  • Not primary alignment (256)")

        # Interpret CIGAR
        print(f"\nCIGAR string: {first_read[5]}")
        print("Interpretation:")
        cigar = first_read[5]
        ops = re.findall(r'(\d+)([MIDNSHP=X])', cigar)
        for length, op in ops:
            length = int(length)
            if op == 'M': print(f"  • {length}M: {length} bases aligned (match/mismatch)")
            elif op == '=': print(f"  • {length}=: {length} bases exact match")
            elif op == 'X': print(f"  • {length}X: {length} bases mismatch")
            elif op == 'I': print(f"  • {length}I: {length} bases insertion in read")
            elif op == 'D': print(f"  • {length}D: {length} bases deletion in reference")
            elif op == 'S': print(f"  • {length}S: {length} bases soft clipped")
    else:
        print("No alignment lines found in SAM file")
```

```python
#  Part C - Convert SAM to BAM and Index


print("="*60)
print("PART C: SAM TO BAM CONVERSION")
print("="*60)

sam_file = "Project1/mapping_output/aligned_reads.sam"
bam_file = "Project1/mapping_output/aligned_reads.bam"
sorted_bam = "Project1/mapping_output/aligned_reads_sorted.bam"

if not os.path.exists(sam_file):
    print(" SAM file not found!")
else:
    print(f"SAM file size: {os.path.getsize(sam_file) / (1024**3):.2f} GB")
    print("\nThis will take a few minutes...\n")

    # Step 1: SAM to BAM
    print("1. Converting SAM to BAM...")
    !samtools view -bS {sam_file} > {bam_file}

    # Step 2: Sort BAM
    print("2. Sorting BAM file...")
    !samtools sort {bam_file} -o {sorted_bam}

    # Step 3: Index BAM
    print("3. Indexing BAM file...")
    !samtools index {sorted_bam}

    # Verify
    print("\n4. Verifying files:")
    !ls -lh {bam_file} {sorted_bam} {sorted_bam}.bai

    print("\n✓ SAM to BAM conversion, sorting, and indexing completed!")
```

```python
#  Part C - Mapping Statistics and Read Depth
"""
Step C5: Percentage of mapped reads
Step C6: Read depth at all positions and mean
"""

print("="*60)
print("PART C: MAPPING STATISTICS AND READ DEPTH")
print("="*60)

sorted_bam = "Project1/mapping_output/aligned_reads_sorted.bam"

if not os.path.exists(sorted_bam):
    print(" Sorted BAM file not found! Run CELL 16 first.")
else:
    # Step C5: Mapping statistics with flagstat
    print("\n" + "="*40)
    print("STEP C5: MAPPING STATISTICS")
    print("="*40)
    !samtools flagstat {sorted_bam}

    # Extract mapping percentage
    result = !samtools flagstat {sorted_bam}
    for line in result:
        if "mapped (" in line and "%" in line:
            percentage = line.split('%')[0].split('(')[-1]
            print(f"\n✓ Percentage of mapped reads: {percentage}%")

    # Step C6: Read depth analysis
    print("\n" + "="*40)
    print("STEP C6: READ DEPTH ANALYSIS")
    print("="*40)

    # Calculate depth for all positions
    print("Calculating read depth across all positions...")
    !samtools depth {sorted_bam} > Project1/mapping_output/read_depth.txt

    # Calculate mean depth
    print("\nMean read depth across all positions:")
    !awk '{{sum+=$3; count++}} END {{if(count>0) print "  " sum/count "X"; else print "  No coverage"}}' Project1/mapping_output/read_depth.txt

    # Additional depth statistics
    print("\nDepth statistics:")
    !awk '{{if(NR==1) min=$3; sum+=$3; count++; if($3>max) max=$3; if($3<min) min=$3}} END {{if(count>0) {{print "  Min depth: " min; print "  Max depth: " max; print "  Mean depth: " sum/count; print "  Positions covered: " count}} else print "  No coverage data"}}' Project1/mapping_output/read_depth.txt

    # Save results
    with open("Project1/QC_results/mapping_statistics.txt", "w") as f:
        f.write("PART C: READ MAPPING RESULTS\n")
        f.write("="*50 + "\n\n")
        for line in result:
            f.write(line + "\n")
        f.write(f"\nMean read depth: ")
        !awk '{{sum+=$3; count++}} END {{print sum/count "X"}}' Project1/mapping_output/read_depth.txt >> "Project1/QC_results/mapping_statistics.txt"

    print("\n✓ Results saved to: Project1/QC_results/mapping_statistics.txt")
```
